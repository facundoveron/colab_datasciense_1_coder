# Implementacion de logica python

## Condicionales
En este caso vamos a usar los condicionales para agregar una nueva columna, donde vamos a categorizar de los alojamiento segun su calificacion general.

Para este vamos a usar la columna review_scores_rating.
Vamos a calular su mediana.
Y en base a eso vamos a categorizar:
1. baja
  '< que la media'
2. media
  '= que la media'
3. alta
  '>' que la media





```python
media_review_scores_rating = df['review_scores_rating'].median()
print(f'La media de las review es {media_review_scores_rating}')

df["deseabilidad"] = np.select(
    [
        df['review_scores_rating'] > media_review_scores_rating,
        df['review_scores_rating'] < media_review_scores_rating,
        df['review_scores_rating'] == media_review_scores_rating
    ],
    [
        "alta",
        "baja",
        "media"
    ],
    default="sin_categoria"
)
```

## Funcion lambda para una transformacion atomica
Vamos a aprovechar y vamos a poner todo en mayuscula la columna que creamos para la deseabilidad

```python
df["deseabilidad"] = df["deseabilidad"].apply(lambda x: x.upper())
```
